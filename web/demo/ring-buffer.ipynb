# Ring buffer, step by step

A **fixed-size** buffer that overwrites the oldest sample.

```python
class Ring:
    def __init__(self, n):
        self.buf = [0.0] * n
        self.i = 0
```

```python
r = Ring(4)
for v in (0.1, 0.5, 0.998, 0.44):
    r.buf[r.i] = v; r.i = (r.i + 1) % 4
print('mean', round(sum(r.buf)/4, 3))
```

## Why it matters
- no allocation in the loop
- constant time writes
